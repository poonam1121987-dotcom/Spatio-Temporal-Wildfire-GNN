```python
import os
import torch
import torch.nn as FileNetwork
import numpy as np
import matplotlib.pyplot as plt

# 1. Install PyTorch Geometric (The AI library for Graph Networks)
print("Installing PyTorch Geometric... This takes about 30 seconds.")
!pip install -q torch-geometric

import torch_geometric
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv

# 2. Create simulated Mediterranean Grid Data (Nodes and Wind Edges)
print("Initializing geographic coordinate nodes...")
np.random.seed(42)
num_coordinates = 50 # 50 distinct grid points in the region

# Node features: [Vegetation Dryness (NDVI), Land Temperature, Soil Moisture]
node_features = np.random.rand(num_coordinates, 3)
node_features = torch.tensor(node_features, dtype=torch.float)

# Create structural edges based on wind vectors (connecting adjacent coordinates)
sources = []
targets = []
for i in range(num_coordinates - 1):
    sources.append(i)
    targets.append(i + 1)
    # Wind propagates fire backwards too based on shifting vectors
    if np.random.rand() > 0.4:
        sources.append(i + 1)
        targets.append(i)

edge_index = torch.tensor([sources, targets], dtype=torch.long)

# Labels: 1 = Fire Outbreak, 0 = Safe Zone
labels = torch.randint(0, 2, (num_coordinates,))

# Package into a Graph Object
geo_graph = Data(x=node_features, edge_index=edge_index, y=labels)

# 3. Define the Spatio-Temporal GNN Architecture
class WildfireGNN(torch.nn.Module):
    def __init__(self):
        super(WildfireGNN, self).__init__()
        self.conv1 = GCNConv(3, 16) # Layer 1: Aggregates spatial data from neighbors
        self.conv2 = GCNConv(16, 2) # Layer 2: Evaluates localized fire risk

    def forward(self, data):
        x, edge_index = data.x, data.edge_index
        x = self.conv1(x, edge_index)
        x = torch.relu(x)
        x = self.conv2(x, edge_index)
        return torch.log_softmax(x, dim=1)

# 4. Initialize and Verify the Model
model = WildfireGNN()
print("\n=== SUCCESS ===")
print("Your Graph Neural Network structure is fully built and ready to run!")
print(f"Total Geographic Coordinates Mapped (Nodes): {geo_graph.num_nodes}")
print(f"Total Structural Wind Pathways Configured (Edges): {geo_graph.num_edges}")
```

```python
# 8. Calculate and Plot Feature Importance (Explainable AI)
print("Running Explainable AI diagnostics...")

# Extract weights from the first convolutional layer
weights = model.conv1.lin.weight.detach().numpy()
importance = np.mean(np.abs(weights), axis=0)
importance = (importance / np.sum(importance)) * 100

features = ['Vegetation Dryness (NDVI)', 'Land Temperature', 'Soil Moisture']

# Generate the diagnostic chart
plt.figure(figsize=(8, 4))
colors = ['#2ecc71', '#e74c3c', '#3498db']
plt.barh(features, importance, color=colors, edgecolor='black', height=0.5)
plt.xlabel('Algorithmic Importance Weight (%)')
plt.title('Explainable AI Diagnostics: Wildfire Risk Drivers')
plt.xlim(0, 100)
plt.grid(axis='x', linestyle='--', alpha=0.7)

for index, value in enumerate(importance):
    plt.text(value + 2, index, f"{value:.1f}%", va='center', fontweight='bold')

plt.tight_layout()
plt.show()

print("\n=== DIAGNOSTICS COMPLETE ===")
print("Your Explainable AI diagnostic chart has been rendered successfully!")
```

```python
# 9. Calculate Precision, Recall, and F1-Score
from sklearn.metrics import precision_score, recall_score, f1_score

# Switch model to evaluation mode
model.eval()

with torch.no_grad():
    # Get model prediction probabilities and select class with highest probability
    out = model(geo_graph)
    preds = out.argmax(dim=1).cpu().numpy()
    targets = geo_graph.y.cpu().numpy()

# Calculate classification metrics
precision = precision_score(targets, preds, zero_division=0)
recall = recall_score(targets, preds, zero_division=0)
f1 = f1_score(targets, preds, zero_division=0)

print("=== DETAILED PERFORMANCE METRICS ===")
print(f"Precision : {precision:.3f}")
print(f"Recall    : {recall:.3f}")
print(f"F1-Score  : {f1:.3f}")
```

```python
import pandas as pd
import numpy as np
import torch
from torch_geometric.data import Data
import kagglehub
import os

# 1. Download dataset automatically from Kaggle
path = kagglehub.dataset_download("nitinchoudhary012/algerian-forest-fires-dataset")

# Locate CSV file
csv_file = [os.path.join(path, f) for f in os.listdir(path) if f.endswith('.csv')][0]

# 2. Load CSV directly (NO skiprows needed)
df = pd.read_csv(csv_file)

# Clean up column spaces (e.g. 'Classes  ' -> 'Classes')
df.columns = df.columns.str.strip()

# 3. Filter out non-numeric/header rows
df = df[pd.to_numeric(df['Temperature'], errors='coerce').notnull()].copy()

# 4. Extract climate features (Temperature, Relative Humidity, Wind Speed)
features = df[['Temperature', 'RH', 'Ws']].astype(float).values

# 5. Extract target labels (1 = fire, 0 = no fire)
labels = (df['Classes'].str.contains('fire', case=False, na=False)).astype(int).values

# 6. Convert to PyTorch Tensors
node_features = torch.tensor(features, dtype=torch.float)
labels_tensor = torch.tensor(labels, dtype=torch.long)
num_coordinates = node_features.shape[0]

# 7. Re-build spatial graph edges
sources, targets = [], []
for i in range(num_coordinates - 1):
    sources.append(i)
    targets.append(i + 1)
    if np.random.rand() > 0.4:
        sources.append(i + 1)
        targets.append(i)

edge_index = torch.tensor([sources, targets], dtype=torch.long)

# 8. Package into Graph Object
geo_graph = Data(x=node_features, edge_index=edge_index, y=labels_tensor)

print(f"=== SUCCESS ===")
print(f"Loaded {num_coordinates} REAL climate data rows into your GNN graph!")
```

```python
import torch
import torch.nn as nn

# Re-instantiate model
model = WildfireGNN()

# Higher learning rate to break the loss plateau
optimizer = torch.optim.Adam(model.parameters(), lr=0.05, weight_decay=1e-4)
criterion = nn.CrossEntropyLoss()

model.train()
for epoch in range(1, 601):
    optimizer.zero_grad()

    # Forward pass
    out = model(geo_graph)

    # Calculate loss and update weights
    loss = criterion(out, geo_graph.y)
    loss.backward()
    optimizer.step()

    if epoch % 20 == 0 or epoch == 1:
        print(f"Iteration {epoch:03d} | Computational Loss: {loss.item():.4f}")

# Calculate training accuracy
preds = out.argmax(dim=1)
acc = (preds == geo_graph.y).sum().item() / geo_graph.y.size(0)
print("\n=== TRAINING COMPLETE ===")
print(f"Your model's spatial forecasting accuracy: {acc * 100:.1f}%")
```
